## Etape 1

```java
import java.util.Random;

Random hasard = new Random();

// Tirage d'un nombre entre 0 et 5, auquel on ajoute 1 pour obtenir 1 à 6
int alea = hasard.nextInt(6) + 1;

// Affichage du résultat
System.out.println(alea);
```

## Etape 2

```java
// On reprend la valeur 'alea' calculée à l'étape 1.
// On initialise les 4 variables avec un espace ' ' par défaut.
char u = ' ';
char d = ' ';
char q = ' ';
char s = ' ';

// Règle pour u (un) : '*' si le nombre est impair (1, 3, 5)
// Le symbole % (modulo) donne le reste de la division. Si le reste par 2 n'est pas 0, c'est impair.
if (alea % 2 != 0) { 
    u = '*'; 
}

// Règle pour d (deux) : '*' si 2 ou plus
if (alea >= 2) { 
    d = '*'; 
}

// Règle pour q (quatre) : '*' si 4 ou plus
if (alea >= 4) { 
    q = '*'; 
}

// Règle pour s (six) : '*' si 6 (l'énoncé disait 2, mais c'est bien 6 pour les cotés latéraux)
if (alea >= 6) { 
    s = '*'; 
}

// Affichage final selon le schéma q d / s u s / d q
System.out.println("+-------+");
System.out.println("| " + q + "   " + d + " |");
System.out.println("| " + s + " " + u + " " + s + " |");
System.out.println("| " + d + "   " + q + " |");
System.out.println("+-------+");
```

## Etape 3

```java
String texte = "";

switch (alea) {
    case 1:  texte = "Un";
             break;
    case 2:  texte = "Deux";
             break;
    case 3:  texte = "Trois";
             break;
    case 4:  texte = "Quatre";
             break;
    case 5:  texte = "Cinq";
             break;
    case 6:  texte = "Six";
             break;
}

System.out.println("Résultat : " + texte + " !");
```

## Etape 4

```java
// Le symbole % (modulo) calcule le reste de la division par 2.
// Si le reste est 0 (ex: 4/2 = 2 reste 0), c'est PAIR.
// Sinon, c'est IMPAIR.

if (alea % 2 == 0) {
    System.out.println("Bonus : Le chiffre " + alea + " est PAIR !");
} else {
    System.out.println("Bonus : Le chiffre " + alea + " est IMPAIR !");
}
```
